### ***The calculation of dispersion measures (Range, Standar Deviation, Variance, Interquartile Range (IQR), Coefficient of variation (CV)***

- ***Temperatura***

```python
print("---------- Medidas de Dispersion ----------")
print("-------- Temperatura del Invernadero --------")

# Rango. La diferencia entre el valor máximo y el valor mínimo en un conjunto de datos
print(f" Rango de la temperatura del invernadero: {df['Temperatura'].max() - df['Temperatura'].min()}")

# La desviación estándar es una medida de la dispersión promedio de los valores con respecto a la media.
print(f" Desviacion Estandar de la temperatura del invernadero: {df['Temperatura'].std()}")

# La varianza es el cuadrado de la desviación estándar.
print(f" Varianza de la temperatura del invernadero: {df['Temperatura'].var()}")

# El rango intercuartil es la diferencia entre el tercer cuartil (Q3) y el primer cuartil (Q1) en un conjunto de datos ordenados.
iqr = df['Temperatura'].quantile(0.75) - df['Temperatura'].quantile(0.25)
print(f" RIC de la temperatura en el invernadero: {iqr}")

# El coeficiente de variación es una medida de la dispersión relativa de los datos. Se obtiene dividiendo la desviación estándar por la media y multiplicando el resultado por 100.
cv = df['Temperatura'].std(ddof=0) / df['Temperatura'].mean()
cv2 = ss.variation(df['Temperatura'])
assert(cv == cv2)

print(f" Coeficiente de Variacion de la temperatura en el invernadero: {cv2}")
```
